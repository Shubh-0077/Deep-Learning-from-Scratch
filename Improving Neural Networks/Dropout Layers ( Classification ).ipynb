
# Dropout in ANN — Classification

## 1. What is Dropout?

**Dropout** is a **regularization technique** used to reduce overfitting.

```python
Dropout(0.2)
```

→ Approximately **20% of neurons are randomly deactivated** during each training iteration.

It prevents **co-adaptation** and encourages the network to learn robust features.

> Dropout is active during training and disabled during inference/testing.

---

## 2. Classification Experiment

### Architecture

```text
Input
  ↓
Dense(128, ReLU)
  ↓
Dense(128, ReLU)
  ↓
Dense(1, Sigmoid)
```

The **Sigmoid** output is used for binary classification.

### Without Dropout — $p=0$

The model strongly overfits.

- Training Accuracy ≈ **92%**
- Validation Accuracy ≈ **74%**
- Decision boundary becomes **jagged and complex**.

```text
High Training Accuracy
        +
Low Validation Accuracy
        ↓
    Overfitting
```

---

## 3. Effect of Dropout

### $p=0.2$

- Decision boundary becomes somewhat smoother.
- Overfitting decreases.
- Validation loss gap becomes smaller.
- Some overfitting may still remain.

### $p=0.5$

- Decision boundary becomes **smooth and continuous**.
- Less sensitive to individual outliers.
- Better represents the general data distribution.
- Validation accuracy increases to approximately **83%+** in the experiment.

---

## 4. Dropout Rate Comparison

| Dropout | Decision Boundary | Result |
|---|---|---|
| **$p=0$** | Very jagged | Severe overfitting |
| **$p=0.2$** | Somewhat smoother | Moderate overfitting |
| **$p=0.5$** | Smooth/general | Better regularization |
| **Very high $p$** | Too simple | Underfitting |

$$
\text{Low Dropout} \rightarrow \text{Overfitting}
$$

$$
\text{Moderate Dropout} \rightarrow \text{Better Generalization}
$$

$$
\text{High Dropout} \rightarrow \text{Underfitting}
$$

---

## 5. Why Does Dropout Help?

### Without Dropout

```text
Full Model Capacity
        ↓
Memorizes Training Data
        ↓
Complex Decision Boundary
        ↓
Overfitting
```

### With Dropout

```text
Random Neurons Deactivated
        ↓
Less Neuron Dependence
        ↓
Robust Features
        ↓
Smoother Decision Boundary
        ↓
Better Generalization
```

---

## 6. How to Choose Dropout?

- **Overfitting** → Increase dropout.
- **Underfitting** → Decrease dropout.
- Practical starting range for many ANN problems: **0.2–0.5**.

### Dropout Placement

It does not need to be used after every layer.

Common approach:

```text
Dense
  ↓
Dense
  ↓
Dropout
  ↓
Output
```

### Other Architectures

- **CNNs:** commonly `0.4–0.5`
- **RNNs/LSTMs:** commonly `0.1–0.5`

These are **guidelines**, not fixed values.

---

## 7. Advantages & Drawbacks

### Advantages

- Reduces overfitting.
- Improves generalization.
- Produces smoother decision boundaries.
- Prevents excessive neuron co-adaptation.

### Drawbacks

- Can slow convergence.
- Makes training dynamics harder to analyze because the active network changes every iteration.

---

## 8. Key Takeaways

- Dropout = **regularization technique**.
- `Dropout(p)` randomly deactivates approximately **$p$ fraction of neurons during training**.
- **$p=0$ → higher overfitting risk.**
- **$p=0.2-0.5$ → useful regularization.**
- **Very high $p$ → underfitting.**
- If overfitting → **increase dropout**.
- If underfitting → **decrease dropout**.
- Focus on **validation/test performance**, not just training accuracy.
```

# Dropout  

````python
# Dropout is 50%

model_2 = Sequential([
    Input(shape=(2,)), 
    Dense(128, activation="relu"),
    Dropout(0.5), 
    Dense(128, activation="relu"),
    Dropout(0.5),
    Dense(1, activation="sigmoid")
])

adam = Adam(learning_rate=0.01)
model_2.compile(loss='binary_crossentropy', optimizer=adam, metrics=['accuracy'])

drop_history = model_2.fit(x, y, epochs=500, validation_split = 0.2,verbose=1)
````
